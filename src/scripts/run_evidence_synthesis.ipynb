```python
import os
import json
from pprint import pprint

from dotenv import load_dotenv

load_dotenv()
```

```python
from research_agent.agents.evidence_synthesis import EvidenceSynthesisAgent

agent = (
        EvidenceSynthesisAgent(
            model_name=os.environ.get("OPENAI_MODEL_NAME"),
            openai_key=os.environ.get("OPENAI_API_KEY"),
        )
        .build_graph()
        .compile()
    )
```

```python
# Load question formulation stub.
with open(
    "/Users/deepankersingh/Projects/research-agent/test/stubs/output_question_formulation.json",
    "r",
) as fp:
    output_question_formulation = json.load(fp)


# Load retrievals stub.
with open(
    "/Users/deepankersingh/Projects/research-agent/test/stubs/output_retriever.json",
    "r",
) as fp:
    output_retriever = json.load(fp)


# Prepare input state.
result = agent.invoke(
    {
        "formulated_question": output_question_formulation,
        "filtered_retrievals": output_retriever.get("filtered_retrievals"),
        "paper_profiles": []
    }
)
```

```python
pprint(result.get("paper_profiles"))
```

```python
# Save results.
from langchain_core.load.dump import dumps
from pydantic.v1.json import pydantic_encoder

# clean_dict = json.loads(dumps(result))
with open(
    "/Users/deepankersingh/Projects/research-agent/test/stubs/output_evidence_synthesis.json",
    "w",
) as fp:
    json.dump({
        "paper_profiles": result["paper_profiles"],
        "filtered_retrievals": result["filtered_retrievals"]
    }, fp, indent=4, ensure_ascii=False)
```
